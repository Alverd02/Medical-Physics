```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
import os
from skimage.metrics import mean_squared_error
from skimage.metrics import structural_similarity
from scipy.stats import pearsonr
```

```python
mapa=nib.load("/content/drive/MyDrive/reconstruccio/receptors_reg.nii")
mapa=np.uint8(mapa.get_fdata())
activitat=nib.load('/content/drive/MyDrive/reconstruccio/act.img')
act = activitat.get_fdata()[:,:,:,0]
diccionari={1:'Caudat esquerre', 2:'Putamen esquerre',3:'Caudat dret',4:'Putamen dret',5:'Cerebel'}

primaris='/content/drive/MyDrive/reconstruccio/rec1'
totals='/content/drive/MyDrive/reconstruccio/rec3'


```

```python
def get_sur(act,mapa,ref=5):
  """
  act: Imatge de la que volem obtenir el SUR
  mapa: Mapa de segmentació
  ref=5: Index del mapa de segmentacio que agafem com a referencia (B)
  vms
  """
  SURarr=np.empty(ref-1)
  act=act/act.max()
  mask=mapa==ref
  B=np.sum(act[mask])/np.sum(mask)
  for i in range(1,mapa.max()+1):
    if i==ref:
      continue

    mask=mapa==i
    S=np.sum(act[mask])/np.sum(mask)
    SUR=(S-B)/B
    SURarr[i-1]=SUR

  return SURarr

```

```python
SUR_teoric=get_sur(act,mapa)
SUR_teoric
```

```python

os.chdir(primaris)
freqs=['002','010','030','050','070','090']
primari_SUR=np.empty((len(freqs),SUR_teoric.size))
for i,freq in enumerate(freqs):
  file_path='reconstruccio1_'+freq+'.img'
  recons=nib.load(file_path)
  rec = recons.get_fdata()[:,:,:,0]
  SURarr=get_sur(rec,mapa)
  primari_SUR[i,:]=SURarr
```

```python
os.chdir(totals)
freqs=['002','010','030','050','070','090']
totals_SUR=np.empty((len(freqs),SUR_teoric.size))
for i,freq in enumerate(freqs):
  file_path='reconstruccio3_'+freq+'.img'
  recons=nib.load(file_path)
  rec = recons.get_fdata()[:,:,:,0]
  SURarr=get_sur(rec,mapa)
  totals_SUR[i,:]=SURarr

```
